```python

# LATIHAN 1: EKSPLORASI DATA & STATISTIKA DESKRIPTIF

import pandas as pd
import glob
import os

# Membaca dataset
files = glob.glob('/content/*.csv')

if not files:
    raise FileNotFoundError("Upload file CSV terlebih dahulu!")

file_path = next(
    (f for f in files if 'penjualan_nusantaramart' in os.path.basename(f).lower()),
    files[0]
)

df = pd.read_csv(file_path)

# Menampilkan 5 data pertama
print("5 DATA PERTAMA")
print(df.head())

# Informasi tipe data dan jumlah data kosong
print("\nINFORMASI DATA")
df.info()

# Statistik deskriptif
print("\nSTATISTIK DESKRIPTIF")
print(df.describe())

# Rata-rata total pembelian per kategori
print("\nRATA-RATA TOTAL PER KATEGORI")
rata_kategori = df.groupby('kategori_produk')['total'].mean()
print(rata_kategori.sort_values(ascending=False))

# Mengecek data kosong
print("\nJUMLAH DATA KOSONG")
print(df.isnull().sum())

# Membandingkan mean dan median
print("\nMEAN DAN MEDIAN")
print("Mean:", df['total'].mean())
print("Median:", df['total'].median())
```

```python

# LATIHAN 2: VISUALISASI PERBANDINGAN DENGAN BAR CHART

import matplotlib.pyplot as plt
import seaborn as sns

# Menghitung total penjualan per kategori
total_kategori = (
    df.groupby('kategori_produk')['total']
    .sum()
    .sort_values(ascending=False)
)

print("TOTAL PENJUALAN PER KATEGORI")
print(total_kategori)

# Membuat Bar Chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    palette='Blues_r'
)

plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')

plt.tight_layout()
plt.show()

# Menampilkan kategori tertinggi dan terendah
print("\nHASIL ANALISIS")
print("Kategori penjualan tertinggi:", total_kategori.idxmax())
print("Total penjualan tertinggi:", total_kategori.max())

print("Kategori penjualan terendah:", total_kategori.idxmin())
print("Total penjualan terendah:", total_kategori.min())
```

```python

# LATIHAN 3: VISUALISASI TREN DENGAN LINE CHART

import pandas as pd
import matplotlib.pyplot as plt

# Mengubah kolom tanggal menjadi format datetime
df['tanggal'] = pd.to_datetime(df['tanggal'])

# Membuat kolom bulan
df['bulan'] = df['tanggal'].dt.to_period('M')

# Menghitung total penjualan setiap bulan
tren_bulanan = df.groupby('bulan')['total'].sum().sort_index()

print("TOTAL PENJUALAN PER BULAN")
print(tren_bulanan)

# Membuat Line Chart
plt.figure(figsize=(8, 5))

plt.plot(
    tren_bulanan.index.astype(str),
    tren_bulanan.values,
    marker='o',
    color='crimson'
)

plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# Menampilkan bulan dengan penjualan tertinggi
print("\nHASIL ANALISIS")

bulan_tertinggi = tren_bulanan.idxmax()
penjualan_tertinggi = tren_bulanan.max()

print("Bulan dengan penjualan tertinggi:", bulan_tertinggi)
print("Total penjualan tertinggi:", penjualan_tertinggi)

# Melihat perubahan penjualan setiap bulan
perubahan = tren_bulanan.diff()

print("\nPERUBAHAN PENJUALAN SETIAP BULAN")
print(perubahan)
```

```python

# LATIHAN 4: VISUALISASI HUBUNGAN DENGAN SCATTER PLOT

import matplotlib.pyplot as plt
import seaborn as sns

# Membuat Scatter Plot
plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='jumlah',
    y='total',
    hue='kategori_produk'
)

plt.title('Hubungan Jumlah Pembelian dengan Total Nilai Transaksi')
plt.xlabel('Jumlah Barang')
plt.ylabel('Total (Rp)')

plt.tight_layout()
plt.show()

# Menghitung korelasi jumlah dan total
print("KORELASI JUMLAH DAN TOTAL")
print(df['jumlah'].corr(df['total']))

# Menampilkan data dengan total transaksi tertinggi
print("\nDATA TRANSAKSI TERTINGGI")
print(df.nlargest(5, 'total')[
    ['kategori_produk', 'jumlah', 'total']
])
```

```python

# LATIHAN 6: DASHBOARD SEDERHANA

import matplotlib.pyplot as plt
import seaborn as sns

# Membuat dashboard dengan 4 grafik
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Grafik 1: Total penjualan per kategori
sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    ax=axes[0, 0],
    palette='Blues_r'
)
axes[0, 0].set_title('Total Penjualan per Kategori')
axes[0, 0].set_xlabel('Total Penjualan (Rp)')
axes[0, 0].set_ylabel('Kategori Produk')

# Grafik 2: Tren penjualan bulanan
axes[0, 1].plot(
    tren_bulanan.index.astype(str),
    tren_bulanan.values,
    marker='o'
)
axes[0, 1].set_title('Tren Penjualan Bulanan')
axes[0, 1].set_xlabel('Bulan')
axes[0, 1].set_ylabel('Total Penjualan (Rp)')
axes[0, 1].tick_params(axis='x', rotation=45)

# Grafik 3: Distribusi rating pelanggan
sns.histplot(
    data=df,
    x='rating_pelanggan',
    bins=5,
    ax=axes[1, 0]
)
axes[1, 0].set_title('Distribusi Rating Pelanggan')
axes[1, 0].set_xlabel('Rating Pelanggan')

# Grafik 4: Sebaran jumlah barang per kategori
sns.boxplot(
    data=df,
    x='kategori_produk',
    y='jumlah',
    ax=axes[1, 1]
)
axes[1, 1].set_title('Sebaran Jumlah per Kategori')
axes[1, 1].set_xlabel('Kategori Produk')
axes[1, 1].set_ylabel('Jumlah Barang')
axes[1, 1].tick_params(axis='x', rotation=45)

# Merapikan tampilan dashboard
plt.tight_layout()
plt.show()
```
